```python
# Part B – [Geometric mean]
import numpy as np
## 1. Formula and explanation

# Geometric mean
# \left(\prod_{i=1}^{N} x_i\right)^{1/N}
# It computes the average of quantities that multiply.

## 2. Loop implementation
def Geometric_mean(counts):
  total = 1
  for value in counts:
    total = total * value
  return total ** (1/len(counts))




## 3. NumPy implementation

def NumPy_Geometric_mean(counts): return np.prod(counts, dtype=float) ** (1 / len(counts))

## 4. Comparison
counts = [
    3473, 2876, 2480, 2445, 1524, 1980, 1237, 3635, 2060, 2386,
    2562, 2109, 1915, 2189, 3094, 1974, 2407, 2653, 2374, 1940,
    1871, 2456, 2140, 2207, 4757, 2882, 3834, 2512, 1663, 2023
]

print("Loop implementation        ", Geometric_mean(counts))
print("NumPy implementation       ", NumPy_Geometric_mean(counts))

```
